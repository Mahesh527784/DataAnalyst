```python
import pandas as pd
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import linkage, dendrogram, fcluster
from matplotlib.lines import Line2D

# Numeric data for clustering
X = data.select_dtypes(include='number')

# Hierarchical clustering
link = linkage(X, method='complete', metric='euclidean')
clusters = fcluster(link, 5, criterion='maxclust')
data['Cluster'] = clusters

# Dendrogram with color threshold
plt.figure(figsize=(18, 8))
dendro = dendrogram(link, labels=data.index.astype(str),
                    leaf_rotation=90, leaf_font_size=6,
                    color_threshold=0.7 * max(link[:, 2]))

# Add legend for cluster colors
colors = list(set(dendro['color_list']))
legend = [Line2D([0], [0], color=c, lw=2, label=f'Cluster {i+1}')
          for i, c in enumerate(colors)]
plt.legend(handles=legend, title='Clusters')
plt.title("University Clusters (Colored Dendrogram)")
plt.xlabel("University")
plt.ylabel("Distance")
plt.tight_layout()
plt.show()


```
