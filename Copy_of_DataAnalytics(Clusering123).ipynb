**Step 1:** Generate clusters using Agglomerative Clustering

```python
# Set number of clusters (based on dendrogram observation)
num_clusters = 4

# Apply hierarchical clustering
hc = AgglomerativeClustering(n_clusters=num_clusters, linkage='complete', metric='euclidean')
y_hc = hc.fit_predict(X_scaled)

# Print assigned cluster labels
print("Cluster Labels:\n", y_hc)

```

**Step 2:** Add cluster info to your original data

```python
# Avoid SettingWithCopyWarning
data.loc[:, 'Cluster'] = y_hc

# Print cluster-wise means only for numeric columns
numeric_cols = data.select_dtypes(include=[np.number]).columns
print(data.groupby('Cluster')[numeric_cols].mean())

```
